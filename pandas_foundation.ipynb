## Part 1 — The two core objects
Before touching the Titanic data, build each object by hand so you can see what they are. A **Series** is one labelled column; a **DataFrame** is a whole table.

```python
# Lesson: pandas has two core objects.
#   Series    -> a single labelled column of data
#   DataFrame -> a whole table (columns that share a row index)
# Try it:
#   make a Series from [10, 20, 30] and print it
#   make a DataFrame from {'name': ['Ann', 'Bo'], 'age': [30, 25]} and show it
s = pd.Series([10, 20, 30])
print(s)
# 0    10
# 1    20
# 2    30
# dtype: int64

people = pd.DataFrame({'name': ['Ann', 'Bo'], 'age': [30, 25]})
people
```

## Part 2 — First look at a DataFrame
Whenever you load data, the first move is always to size it up before you trust it.

```python
# Lesson: quick ways to size up a DataFrame:
#   df.shape   -> (rows, columns)     df.columns -> the column names
#   df.head(n) -> the first n rows    df.info()  -> columns, types, missing counts
# Try it (use `df` from Setup): print df.shape and the column names,
# show the first 5 rows, and call df.info().

# Part 2
print(df.shape)              # (891, 8)
print(df.columns.tolist())
# ['survived', 'pclass', 'sex', 'age', 'sibsp', 'parch', 'fare', 'embarked']
df.info()                    # types + non-null counts (spot the missing ages)
df.head()
```

## Part 3 — Grab columns
A single column pulled out of a DataFrame *is* a Series. Ask for several at once and you get a smaller DataFrame back.

```python
# Lesson: df['col'] gives ONE column (a Series).
#         df[['a', 'b']] gives SEVERAL columns (a DataFrame) - note the double [[ ]].
# Try it (use `df`):
#   pull out the 'age' column on its own
#   pull out 'sex' and 'age' together

age = df['age']              # one column -> a Series
print(type(age))             # <class 'pandas.core.series.Series'>
df[['sex', 'age']].head()
```

## Part 4 — Add & drop columns
You build new columns out of existing ones all the time. Adding is just assignment; removing uses `.drop()`.

```python
# Lesson: assign to a new name to add a column; use .drop(columns=[...]) to remove one.
#   df['family_size'] = df['sibsp'] + df['parch']
#   df.drop(columns=['family_size'])   # returns a copy without that column
# Try it (use `df`): add a 'family_size' column (sibsp + parch), show the first
# few rows to check it, then drop it again.

df['family_size'] = df['sibsp'] + df['parch']
print(df[['sibsp', 'parch', 'family_size']].head())
df = df.drop(columns=['family_size'])
print(df.columns.tolist())
```

## Part 5 — Select rows: loc vs iloc
Two labels for two ideas: `iloc` picks by **position** (like list indexing), `loc` picks by **label** (the index and column names).

```python
# Lesson:
#   df.iloc[0]        -> the first row, BY POSITION
#   df.iloc[:5]       -> the first five rows
#   df.loc[0, 'age']  -> a single value BY LABEL (row index 0, column 'age')
# Try it (use `df`): show the very first row, then just the 'age' of that first row.

print(df.iloc[0])            # the whole first row, as a Series
print(df.loc[0, 'age'])
```

## Part 6 — Filter with conditions *(bonus)*
This is the one you'll reach for constantly. A condition makes a True/False mask (exactly like NumPy), and you use it to keep only the rows that pass.

```python
# Lesson: a condition builds a True/False mask, and you filter rows with it:
#   df[df['age'] > 60]                                  -> passengers over 60
#   df[(df['sex'] == 'female') & (df['pclass'] == 1)]   -> combine with & / |
# Try it (use `df`):
#   select passengers younger than 12
#   select first-class female passengers
#   bonus: how many survived? try df['survived'].sum(), or df['sex'].value_counts()

# Part 6
kids = df[df['age'] < 12]
print(kids.shape)            # (68, 8)  -> 68 passengers under 12

first_class_women = df[(df['sex'] == 'female') & (df['pclass'] == 1)]
print(first_class_women.shape)   # (94, 8)

print(df['survived'].sum())      # 342 survived
print(df['sex'].value_counts())
# male      577
# female    314
```
